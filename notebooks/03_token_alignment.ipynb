```python
## 2. Load FUNSD and LayoutLMv3
from src.data import load_funsd
from src.processor import build_processor

dataset = load_funsd()
processor = build_processor()

print("Environment ready.")
```

```python
## 3. Compare Words and Subword Tokens
example = dataset["train"][0]

encoding = processor(
    images=example["image"].convert("RGB"),
    text=example["words"],
    boxes=example["bboxes"],
    word_labels=example["ner_tags"],
    truncation=True,
    max_length=512,
)

word_ids = encoding.word_ids()

valid_word_ids = [
    word_id
    for word_id in word_ids
    if word_id is not None
]

print("Original words:", len(example["words"]))
print("Encoded token positions:", len(word_ids))
print(
    "Unique represented words:",
    len(set(valid_word_ids)),
)
```

```python
## 4. Inspect Word-to-Token Mapping
tokens = processor.tokenizer.convert_ids_to_tokens(
    encoding["input_ids"]
)

for index in range(min(80, len(tokens))):
    print(
        f"{index:3d}",
        f"{str(word_ids[index]):>5}",
        tokens[index],
    )
```

```python
## 5. Measure Subword Expansion
from collections import Counter

counts = Counter(valid_word_ids)

single_token_words = sum(
    count == 1
    for count in counts.values()
)

multi_token_words = sum(
    count > 1
    for count in counts.values()
)

print("Single-token words:", single_token_words)
print("Multi-token words:", multi_token_words)

print(
    "Multi-token percentage:",
    round(
        multi_token_words / len(counts) * 100,
        2,
    ),
    "%",
)
```

```python
## 6. Measure Truncation Across FUNSD
truncated = 0
complete = 0

for document in dataset["train"]:
    encoded = processor(
        images=document["image"].convert("RGB"),
        text=document["words"],
        boxes=document["bboxes"],
        word_labels=document["ner_tags"],
        truncation=True,
        max_length=512,
    )

    represented = {
        word_id
        for word_id in encoded.word_ids()
        if word_id is not None
    }

    if len(represented) < len(document["words"]):
        truncated += 1
    else:
        complete += 1

print("Complete documents:", complete)
print("Truncated documents:", truncated)
print(
    "Truncation rate:",
    round(
        truncated / len(dataset["train"]) * 100,
        2,
    ),
    "%",
)
```
